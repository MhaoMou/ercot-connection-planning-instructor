# Objective: inspect a scenario fixture

This fixture-first example inventories public-shaped files and demonstrates the checks a student should run before using an instructor-approved ACTIVSg2000 scenario.

```python
from pathlib import Path
import json, os, platform, sys

TUTORIAL_ROOT = Path(os.environ.get('TUTORIAL_ROOT', Path.cwd())).resolve()
OUTPUT_ROOT = Path(os.environ.get('TUTORIAL_OUTPUT_ROOT', TUTORIAL_ROOT / 'outputs')).resolve()
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
FIXTURE_ROOT = OUTPUT_ROOT / 'fixtures'
FIXTURE_ROOT.mkdir(parents=True, exist_ok=True)
```

```python
print('Environment report')
print({'python': sys.version.split()[0], 'platform': platform.platform(), 'tutorial_root': str(TUTORIAL_ROOT), 'fixture_first': True})
```

```python
buses = [{'bus_id': '1001', 'load_mw': 35.0}, {'bus_id': '1002', 'load_mw': 20.0}, {'bus_id': '1003', 'load_mw': 10.0}]
branches = [{'from_bus': '1001', 'to_bus': '1002', 'reactance_pu': 0.1}, {'from_bus': '1002', 'to_bus': '1003', 'reactance_pu': 0.1}]
manifest = {'dataset': 'small-public-shaped-fixture', 'validation_status': 'not_run', 'scenario_horizon': 'one-hour half-year scenarios', 'geographic_coordinates': False}
bus_ids = {bus['bus_id'] for bus in buses}
assert len(buses) == 3
assert {endpoint for row in branches for endpoint in (row['from_bus'], row['to_bus'])} <= bus_ids
assert manifest['validation_status'] != 'passed'
print('No scenario is claimed validated: fixture manifest status is', manifest['validation_status'])
```

```python
OUTPUT = OUTPUT_ROOT / 'dataset_inspection.json'
summary = {'files': {'bus_fixture.json': {'rows': len(buses), 'columns': sorted(buses[0])}, 'branch_fixture.json': {'rows': len(branches), 'columns': sorted(branches[0])}}, 'duplicate_bus_keys': 0, 'missing_values': 0, 'valid_bus_references': True, 'scenario_horizon': manifest['scenario_horizon'], 'manifest_status': manifest['validation_status'], 'coordinates_available': manifest['geographic_coordinates']}
OUTPUT.write_text(json.dumps(summary, indent=2, sort_keys=True) + '\n', encoding='utf-8')
print('OUTPUT:', OUTPUT)
```
