# Objective: visualize a network fixture

Create deterministic structural, generator-overlay, and demand-summary figures. Geographic coordinates are intentionally absent from this public-shaped fixture.

```python
from pathlib import Path
import os, sys
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

TUTORIAL_ROOT = Path(os.environ.get('TUTORIAL_ROOT', Path.cwd())).resolve()
OUTPUT_ROOT = Path(os.environ.get('TUTORIAL_OUTPUT_ROOT', TUTORIAL_ROOT / 'outputs')).resolve()
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
buses = {'1001': (0.0, 0.0), '1002': (1.0, 0.5), '1003': (2.0, 0.0), '1004': (1.0, -0.8)}
branches = [('1001', '1002'), ('1002', '1003'), ('1002', '1004'), ('1001', '1004')]
generators = {'1001': 80.0, '1003': 30.0}
demand = {'1001': 35.0, '1002': 20.0, '1003': 10.0, '1004': 15.0}
```

```python
print('Environment report')
print({'matplotlib': matplotlib.__version__, 'backend': matplotlib.get_backend(), 'fixture_first': True})
```

```python
assert set(demand) == set(buses)
assert all(left in buses and right in buses for left, right in branches)
assert all(value >= 0 for value in generators.values())
print('Geographic coordinates are unavailable; plots below are structural, not geographic.')
```

```python
OUTPUT = OUTPUT_ROOT / 'network_structure.png'
fig, ax = plt.subplots(figsize=(5, 3.5))
for left, right in branches:
    ax.plot([buses[left][0], buses[right][0]], [buses[left][1], buses[right][1]], color='0.55', linewidth=1.5)
ax.scatter([p[0] for p in buses.values()], [p[1] for p in buses.values()], color='#2563eb', s=70)
for bus, (x, y) in buses.items(): ax.text(x + 0.04, y + 0.04, bus)
ax.set_title('Structural bus and branch fixture'); ax.set_aspect('equal'); fig.tight_layout(); fig.savefig(OUTPUT, dpi=120); plt.close(fig)
print('OUTPUT:', OUTPUT)
```

```python
OUTPUT = OUTPUT_ROOT / 'generator_overlay.png'
fig, ax = plt.subplots(figsize=(5, 3.5))
ax.scatter([p[0] for p in buses.values()], [p[1] for p in buses.values()], color='#94a3b8', s=70, label='bus')
for bus, capacity in generators.items(): ax.scatter(*buses[bus], s=80 + capacity, color='#dc2626', label='generator' if bus == '1001' else None)
for left, right in branches: ax.plot([buses[left][0], buses[right][0]], [buses[left][1], buses[right][1]], color='0.8')
ax.set_title('Generator locations on structural bus layout'); ax.legend(); ax.set_aspect('equal'); fig.tight_layout(); fig.savefig(OUTPUT, dpi=120); plt.close(fig)
print('OUTPUT:', OUTPUT)
```

```python
OUTPUT = OUTPUT_ROOT / 'demand_summary.png'
fig, ax = plt.subplots(figsize=(5, 3)); ax.bar(list(demand), list(demand.values()), color='#16a34a'); ax.set_xlabel('bus_id'); ax.set_ylabel('load (MW)'); ax.set_title('Fixture demand summary'); fig.tight_layout(); fig.savefig(OUTPUT, dpi=120); plt.close(fig)
print('OUTPUT:', OUTPUT)
```
