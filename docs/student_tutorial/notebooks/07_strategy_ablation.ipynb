# Objective: compare controlled strategy ablations

Vary one capacity factor at a time on synthetic candidate plans. These are educational experiments, not optimal strategies and not evidence about another team.

```python
from pathlib import Path
import csv, json, os, sys
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

TUTORIAL_ROOT = Path(os.environ.get('TUTORIAL_ROOT', Path.cwd())).resolve()
OUTPUT_ROOT = Path(os.environ.get('TUTORIAL_OUTPUT_ROOT', TUTORIAL_ROOT / 'outputs')).resolve(); OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
columns = ['team_id', 'site_id', 'bus_id', 'connection_capacity_mw', 'generator_capacity_mw', 'storage_power_mw', 'storage_energy_mwh']
candidates = [('baseline', 40, 20, 0, 0), ('more_connection', 60, 20, 0, 0), ('add_generation', 40, 30, 0, 0), ('add_storage', 40, 20, 5, 20)]
```

```python
print('Environment report')
print({'python': sys.version.split()[0], 'matplotlib': matplotlib.__version__, 'fixture_first': True, 'private_data': False})
```

```python
rows = []
for name, connection, generation, storage_power, storage_energy in candidates:
    assert generation <= connection and storage_energy == 4 * storage_power
    preview_metric = connection + 0.5 * generation + 0.25 * storage_energy
    rows.append({'variant': name, 'team_id': 'fixture-team', 'site_id': 'site-north', 'bus_id': '1001', 'connection_capacity_mw': connection, 'generator_capacity_mw': generation, 'storage_power_mw': storage_power, 'storage_energy_mwh': storage_energy, 'validation_status': 'valid', 'preview_metric': preview_metric})
assert len(rows) == 4 and all(row['validation_status'] == 'valid' for row in rows)
print('Controlled ablation variants:', [row['variant'] for row in rows])
```

```python
OUTPUT = OUTPUT_ROOT / 'strategy_ablation.csv'
fieldnames = columns + ['variant', 'validation_status', 'preview_metric']
with OUTPUT.open('w', newline='', encoding='utf-8') as handle:
    writer = csv.DictWriter(handle, fieldnames=fieldnames); writer.writeheader(); writer.writerows(rows)
print('OUTPUT:', OUTPUT)
```

```python
OUTPUT = OUTPUT_ROOT / 'strategy_ablation.png'
fig, ax = plt.subplots(figsize=(6, 3.5)); ax.bar([row['variant'] for row in rows], [row['preview_metric'] for row in rows], color='#7c3aed'); ax.set_ylabel('fixture preview metric'); ax.set_title('One-factor-at-a-time strategy ablation'); ax.tick_params(axis='x', rotation=25); fig.tight_layout(); fig.savefig(OUTPUT, dpi=120); plt.close(fig)
print('OUTPUT:', OUTPUT)
```
